## plot()- This function is used to plot a trend/ linegraph 
#### 1. When we are having a data which is in ordered form like increasing or decreasing order
#### 2.When the data shows a Trend
#### 3.The data is continuous 
#### for example you are having a data of monthly sales

## Scatter Plots - scatter()
#### When to use - if there is a relationship /pattern in the dataset or you want to see the pattern or relationship in the data
#### it is not necessary that the x axis is ordered
#### these plots are broadly used to show correlation clustering or outliers
#### For example - Hours studying vs Marks obtained by the students

## BarChart - bar()
#### when we want to compare the data 
#### For example - Student Names and their Heights 
#### in this data x-axis is not continuous 


#### Problem Statement
#### 1- X-axis time / ordered ? - month /years/ sequence
####    -> plot()  (line graph )
#### 2- If there is a relation between two variables
####   -> scatter() (Scatter Plots)
#### 3- If you want to compare Categories (names,types,groups) (x axis non-continuous)
####   -> bar() (BarChart)

```python
marks=[23,45,60,65,70,72,75,78,80,82,85,88,90,95,180]
q1=np.quantile(marks,0.25)
q2=np.quantile(marks,0.50)
q3=np.quantile(marks,0.75)

Iqr_val = q3-q1
Iqr_val
lower_boundary = q1-1.5*Iqr_val
upper_boundary = q3+1.5*Iqr_val

median = np.median(marks)
mean = np.mean(marks)
```

```python
box=plt.boxplot(marks,patch_artist=True,showmeans=True,meanprops=dict(marker="D",markerfacecolor="yellow"))
box["boxes"][0].set_facecolor="lightblue"
plt.title("Box Plot")
plt.ylabel("Marks")
plt.xticks([1],["Students"])
plt.grid(axis='y',alpha=0.5)

plt.text(1,150,"Outlier Upper value",fontsize=10,color="red")
plt.text(1,20,"Outlier Lower value",fontsize=10,color="red")
plt.text(1.25,median,"Median (180)",fontsize=10,color="red")
plt.text(1.25,q3,"Q3",fontsize=10,color="red")
plt.text(1.25,q1,"Q1",fontsize=10,color="red")

plt.show()
```
