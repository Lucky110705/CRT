```python
import pandas as pd
import numpy as np
```

```python
comp_A=[480,500,510,520,530,540,550,570,580,590,600,560,610,620,630,60,650,660]
comp_B=[300,320,340,370,390,400,380,400,390,430,440,500,480,540,600,850,1200,1220]
comp_C=[50,100,250,700,750,800,740,750,760,780,790,810,800,810,820,830,840,900]
df=pd.DataFrame({
    "A":comp_A,
    "B":comp_B,
    "C":comp_C 
})
df
```

### For A

```python
a=df["A"].sort_values()
print(a)
q1=np.quantile(a,0.25)
q2=np.quantile(a,0.50)
q3=np.quantile(a,0.75)
print("Q1:",q1)
print("Q2:",q2)
print("Q3:",q3)
iqr=q3-q1
print("IQR:",iqr)
lb = q1-1.5*iqr
ub = q3+1.5*iqr
print("LB:",lb)
print("UB:",ub)

```

```python
outlier = a[(a<lb) | (a>ub)]
print("Outlier:",outlier)
```
